```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector, plot_state_qsphere
import matplotlib.pyplot as plt
import numpy as np
```

```python
qc = QuantumCircuit(1)

# Apply X gate
qc.x(0)

# Display Circuit
print("Quantum Circuit:")
display(qc.draw("mpl"))

# Statevector
state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

# Bloch Sphere
plot_bloch_multivector(state)
plt.show()
```

```python
qc = QuantumCircuit(1)

qc.h(0)
qc.s(0)

print("Quantum Circuit:")
display(qc.draw("mpl"))

state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

plot_bloch_multivector(state)
plt.show()
```

```python
qc = QuantumCircuit(2)

qc.h(0)
qc.h(1)

print("Quantum Circuit:")
display(qc.draw("mpl"))

state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

print("\nProbabilities:")

prob = np.abs(state.data)**2

for i,p in enumerate(prob):
    print(f"|{i:02b}> = {p:.2f}")

print("\nSum =", np.sum(prob))

plot_state_qsphere(state)
plt.show()
```

```python
theta = np.random.uniform(0, np.pi/8)

qc = QuantumCircuit(1)

qc.h(0)
qc.p(theta,0)

print("Random Phase =", theta)

print("Quantum Circuit:")
display(qc.draw("mpl"))

state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

plot_bloch_multivector(state)
plt.show()
```

```python
theta = np.pi/3
phi = np.pi/4

qc = QuantumCircuit(1)

qc.ry(theta,0)
qc.rz(phi,0)

print("Quantum Circuit:")
display(qc.draw("mpl"))

state = Statevector.from_instruction(qc)

print("Statevector:")
print(state)

plot_bloch_multivector(state)
plt.show()
```
